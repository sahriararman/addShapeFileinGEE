```python
import ee
import geemap

# Use 'localhost' mode — bypasses the incompatible OAuth client issue
ee.Authenticate(auth_mode='localhost')
ee.Initialize(project='braided-torch-453121-v4')
print('✅ GEE initialized')
```

```python
import geopandas as gpd
import json

# ── Load study area from local shapefile ──
SHP_PATH = r'H:\gis all\test matuail land\shape file\Jatrabari.shp'

aoi_gdf = gpd.read_file(SHP_PATH)
aoi_gdf = aoi_gdf.to_crs(epsg=4326)

# Convert to GEE geometry (handles multiple features by dissolving)
if len(aoi_gdf) > 1:
    aoi_gdf = aoi_gdf.dissolve()

geojson = json.loads(aoi_gdf.geometry.to_json())
bhairab = ee.Geometry(geojson['features'][0]['geometry'])

print('Source  :', SHP_PATH)
print('CRS     :', aoi_gdf.crs)
print('Bounds  :', aoi_gdf.total_bounds)
print('Features:', len(aoi_gdf))
print('✅ Jatrabari boundary loaded into GEE')

# Quick map preview
minx, miny, maxx, maxy = aoi_gdf.total_bounds  # ← fixed typo here
Map = geemap.Map(center=[(miny+maxy)/2, (minx+maxx)/2], zoom=13)
Map.addLayer(bhairab, {'color': 'red'}, 'Jatrabari')
Map
```
